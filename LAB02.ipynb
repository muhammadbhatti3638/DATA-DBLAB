```python
import pandas as pd
df = pd.read_csv("students_marks - students_marks.csv")
print(df.shape)
```

Lab Task 2: Summarise with pandas

```python
subjects = ["Math", "Stats", "Python", "Databases", "English"]
summary = df.groupby("Section")[subjects].mean().round(2).reset_index()
print(summary)
```

```python
df["Average"] = df[subjects].mean(axis=1).round(2)
print(df.sort_values("Average", ascending=False)[["Name", "Average"]].head(3))
```

Lab Task 3: Write JSON

```python
summary.to_json("summary.json", orient="records")
df.to_json("students.json", orient="records")
print(open("summary.json").read())
```

### Lab Task 4: JavaScript answers questions

```python
%%writefile t4.js
const fs = require('fs');

try {
    const summaryData = JSON.parse(fs.readFileSync('summary.json', 'utf8'));

    // Create a copy and sort descending by Python score
    const sortedSummary = [...summaryData].sort((a, b) => b.Python - a.Python);
    const topPythonSection = sortedSummary[0];

    console.log('Section with the highest average in Python:');
    console.log(`Section: ${topPythonSection.Section}, Python Average: ${topPythonSection.Python}`);
} catch (error) {
    console.error('Error reading or parsing JSON files:', error.message);
}
```

```python
%%writefile t5.js
const fs = require('fs');

try {
    const summaryData = JSON.parse(fs.readFileSync('summary.json', 'utf8'));

    summaryData.forEach(row => {
        console.log(`Section ${row.Section}: Math ${row.Math.toFixed(1)}, Python ${row.Python.toFixed(1)}`);
    });
} catch (error) {
    console.error('Error:', error.message);
}
```

Lab Task 5: Draw the chart with JavaScript

```python
%%writefile chart.js
// data comes from Python (summary.json)
const base = 250; // y of the x-axis
let svg = `<svg width="460" height="300" font-family="Arial">`;
svg += `<text x="230" y="24" text-anchor="middle"` +
 ` font-weight="bold">Average Math by section</text>`;
svg += `<line x1="40" y1="${base}" x2="440" y2="${base}"` +
 ` stroke="black"/>`;
data.forEach((d, i) => {
 const h = d.Math * 2; // 100 marks = 200 px
 const x = 60 + i * 120; // left edge of the bar
 svg += `<rect x="${x}" y="${base - h}" width="80"` +
 ` height="${h}" fill="#0F6E7C"/>`;
 svg += `<text x="${x + 40}" y="${base - h - 6}"` +
 ` text-anchor="middle">${d.Math}</text>`;
 svg += `<text x="${x + 40}" y="${base + 20}"` +
 ` text-anchor="middle">Section ${d.Section}</text>`;
});
svg += `</svg>`;
document.getElementById("chart").innerHTML = svg;
```
